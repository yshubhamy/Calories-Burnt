```python
#import libraries
import numpy as np
import pandas as pd
```

```python
# importing dataset of calories and converting it into pandas data frame
df_calories = pd.read_csv('calories_exercise/calories.csv')
```

```python
df_calories.head()
```

```python
# importing dataset of exercise and converting it into pandas data frame
df_exercise = pd.read_csv("calories_exercise/exercise.csv")
```

```python
df_exercise.head()
```

```python
# joining both the datasets
df = pd.merge(df_exercise, df_calories, on='User_ID', how='inner')
df
```

#### Converting Text data into numerical data

```python
df.replace({"Gender":{"male":0,"female":1}}, inplace=True)
```

Separating Features and Target

```python
X = df.drop(columns=['User_ID',"Calories"], axis=1)
Y = df['Calories']
```

```python
print(X)
```

```python
pip install --upgrade xgboost scikit-learn
```

```python
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

### XGBoost Regressor

```python
#loading the model
model = XGBRegressor()
```

```python
#training the model with X_train
model.fit(X_train, Y_train)
```

### Evaluation

```python
# prediction on train data
train_data_prediction = model.predict(X_train)
```

```python
print(train_data_prediction)
```

```python
# prediction on test data
test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

### Mean Absolute Error

```python
from sklearn import metrics
from sklearn.metrics import r2_score
```

```python
mae = metrics.mean_absolute_error(Y_test, test_data_prediction)
```

```python
print("Mean Absolute Error = ", mae)
```

```python
lossy = r2_score(Y_test, test_data_prediction)
lossy
```

```python
lossy2 = r2_score(Y_train, train_data_prediction)
lossy2
```
